```python
# ==========================================================
# AI-Assisted E-commerce Sales & Customer Analytics
# Data Cleaning - Part 1
# ==========================================================

# ==========================================================
# STEP 1 : Import Required Libraries
# ==========================================================

# Pandas is used for data manipulation and analysis
import pandas as pd

# NumPy is used for numerical operations
import numpy as np

# Display all columns while viewing DataFrames
pd.set_option('display.max_columns', None)

# Display wider output in Jupyter Notebook
pd.set_option('display.width', None)

print("Libraries Imported Successfully!")
```

```python
# ==========================================================
# STEP 2 : Load All Datasets
# ==========================================================

customers = pd.read_csv("olist_customers_dataset.csv")

geolocation = pd.read_csv("olist_geolocation_dataset.csv")

order_items = pd.read_csv("olist_order_items_dataset.csv")

order_payments = pd.read_csv("olist_order_payments_dataset.csv")

order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")

orders = pd.read_csv("olist_orders_dataset.csv")

products = pd.read_csv("olist_products_dataset.csv")

print("All datasets loaded successfully!")
```

```python
# ==========================================================
# STEP 3 : Display First Five Records
# Purpose:
# View sample records from each dataset.
# ==========================================================

print("Customers Dataset")
display(customers.head())

print("Geolocation Dataset")
display(geolocation.head())

print("Order Items Dataset")
display(order_items.head())

print("Order Payments Dataset")
display(order_payments.head())

print("Order Reviews Dataset")
display(order_reviews.head())

print("Orders Dataset")
display(orders.head())

print("Products Dataset")
display(products.head())
```

```python
# ==========================================================
# STEP 4 : Check Dataset Dimensions
# Purpose:
# Display number of rows and columns.
# ==========================================================

print("Customers :", customers.shape)

print("Geolocation :", geolocation.shape)

print("Order Items :", order_items.shape)

print("Order Payments :", order_payments.shape)

print("Order Reviews :", order_reviews.shape)

print("Orders :", orders.shape)

print("Products :", products.shape)
```

```python
# ==========================================================
# STEP 5 : Display Dataset Information
# Purpose:
# Check column names, data types and missing values.
# ==========================================================

print("\nCustomers Dataset Information")
customers.info()

print("\nGeolocation Dataset Information")
geolocation.info()

print("\nOrder Items Dataset Information")
order_items.info()

print("\nOrder Payments Dataset Information")
order_payments.info()

print("\nOrder Reviews Dataset Information")
order_reviews.info()

print("\nOrders Dataset Information")
orders.info()

print("\nProducts Dataset Information")
products.info()
```

```python
# ==========================================================
# STEP 6 : Display Statistical Summary
# Purpose:
# Generate summary statistics for numerical columns.
# ==========================================================

print("\nCustomers")
display(customers.describe(include='all'))

print("\nGeolocation")
display(geolocation.describe(include='all'))

print("\nOrder Items")
display(order_items.describe(include='all'))

print("\nOrder Payments")
display(order_payments.describe(include='all'))

print("\nOrder Reviews")
display(order_reviews.describe(include='all'))

print("\nOrders")
display(orders.describe(include='all'))

print("\nProducts")
display(products.describe(include='all'))
```

```python
# ==========================================================
# STEP 7 : Check Missing Values
# Purpose:
# Identify columns containing null values.
# ==========================================================

print("\nMissing Values - Customers")
print(customers.isnull().sum())

print("\nMissing Values - Geolocation")
print(geolocation.isnull().sum())

print("\nMissing Values - Order Items")
print(order_items.isnull().sum())

print("\nMissing Values - Order Payments")
print(order_payments.isnull().sum())

print("\nMissing Values - Order Reviews")
print(order_reviews.isnull().sum())

print("\nMissing Values - Orders")
print(orders.isnull().sum())

print("\nMissing Values - Products")
print(products.isnull().sum())
```

```python
# ==========================================================
# STEP 8 : Check Duplicate Records
# Purpose:
# Identify duplicate rows in each dataset.
# ==========================================================

print("Duplicate Records")

print("Customers :", customers.duplicated().sum())

print("Geolocation :", geolocation.duplicated().sum())

print("Order Items :", order_items.duplicated().sum())

print("Order Payments :", order_payments.duplicated().sum())

print("Order Reviews :", order_reviews.duplicated().sum())

print("Orders :", orders.duplicated().sum())

print("Products :", products.duplicated().sum())
```

```python
# ==========================================================
# STEP 9 : Remove Duplicate Records
# Purpose:
# Remove duplicate rows if present.
# ==========================================================

customers.drop_duplicates(inplace=True)

geolocation.drop_duplicates(inplace=True)

order_items.drop_duplicates(inplace=True)

order_payments.drop_duplicates(inplace=True)

order_reviews.drop_duplicates(inplace=True)

orders.drop_duplicates(inplace=True)

products.drop_duplicates(inplace=True)

print("Duplicate records removed successfully!")
```

```python
# ==========================================================
# STEP 10 : Verify Dataset Dimensions After Cleaning
# Purpose:
# Confirm updated dataset sizes.
# ==========================================================

print("Customers :", customers.shape)

print("Geolocation :", geolocation.shape)

print("Order Items :", order_items.shape)

print("Order Payments :", order_payments.shape)

print("Order Reviews :", order_reviews.shape)

print("Orders :", orders.shape)

print("Products :", products.shape)
```

```python
# ==========================================================
# STEP 11 : Check Data Types
# Purpose:
# Verify the data type of each column before cleaning.
# ==========================================================

print("Customers Data Types")
display(customers.dtypes)

print("Geolocation Data Types")
display(geolocation.dtypes)

print("Order Items Data Types")
display(order_items.dtypes)

print("Order Payments Data Types")
display(order_payments.dtypes)

print("Order Reviews Data Types")
display(order_reviews.dtypes)

print("Orders Data Types")
display(orders.dtypes)

print("Products Data Types")
display(products.dtypes)
```

```python
# ==========================================================
# STEP 12 : Convert Date Columns
# Purpose:
# Convert object columns containing dates into datetime format.
# ==========================================================

orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

orders["order_approved_at"] = pd.to_datetime(
    orders["order_approved_at"]
)

orders["order_delivered_carrier_date"] = pd.to_datetime(
    orders["order_delivered_carrier_date"]
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"]
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"]
)

order_reviews["review_creation_date"] = pd.to_datetime(
    order_reviews["review_creation_date"]
)

order_reviews["review_answer_timestamp"] = pd.to_datetime(
    order_reviews["review_answer_timestamp"]
)

print("Date columns converted successfully!")
```

```python
# ==========================================================
# STEP 13 : Verify Date Conversion
# ==========================================================

orders.info()

order_reviews.info()
```

```python
# ==========================================================
# STEP 14 : Remove Leading and Trailing Spaces
# Purpose:
# Clean text columns by removing unwanted spaces.
# ==========================================================

customers = customers.apply(
    lambda col: col.str.strip() if col.dtype == "object" else col
)

geolocation = geolocation.apply(
    lambda col: col.str.strip() if col.dtype == "object" else col
)

order_items = order_items.apply(
    lambda col: col.str.strip() if col.dtype == "object" else col
)

order_payments = order_payments.apply(
    lambda col: col.str.strip() if col.dtype == "object" else col
)

order_reviews = order_reviews.apply(
    lambda col: col.str.strip() if col.dtype == "object" else col
)

orders = orders.apply(
    lambda col: col.str.strip() if col.dtype == "object" else col
)

products = products.apply(
    lambda col: col.str.strip() if col.dtype == "object" else col
)

print("Whitespace removed successfully!")
```

```python
# ==========================================================
# STEP 15 : Standardize Text Columns
# Purpose:
# Convert text columns to lowercase for consistency.
# ==========================================================

customers = customers.apply(
    lambda col: col.str.lower() if col.dtype == "object" else col
)

geolocation = geolocation.apply(
    lambda col: col.str.lower() if col.dtype == "object" else col
)

order_items = order_items.apply(
    lambda col: col.str.lower() if col.dtype == "object" else col
)

order_payments = order_payments.apply(
    lambda col: col.str.lower() if col.dtype == "object" else col
)

order_reviews = order_reviews.apply(
    lambda col: col.str.lower() if col.dtype == "object" else col
)

orders = orders.apply(
    lambda col: col.str.lower() if col.dtype == "object" else col
)

products = products.apply(
    lambda col: col.str.lower() if col.dtype == "object" else col
)

print("Text standardized successfully!")
```

```python
# ==========================================================
# STEP 16 : Check Unique Values
# Purpose:
# Identify the number of distinct values in each column.
# ==========================================================

print("Customers")
display(customers.nunique())

print("Geolocation")
display(geolocation.nunique())

print("Order Items")
display(order_items.nunique())

print("Order Payments")
display(order_payments.nunique())

print("Order Reviews")
display(order_reviews.nunique())

print("Orders")
display(orders.nunique())

print("Products")
display(products.nunique())
```

```python
# ==========================================================
# STEP 17: Check for Negative Numeric Values
# Purpose:
# Identify unexpected negative values in numeric columns.
# ==========================================================

datasets = {
    "Order Items": order_items,
    "Order Payments": order_payments,
    "Products": products
}

for dataset_name, df in datasets.items():
    
    print(f"\n{dataset_name}")
    
    numeric_columns = df.select_dtypes(include=np.number).columns
    
    for column in numeric_columns:
        
        negative_values = (df[column] < 0).sum()
        
        print(f"{column}: {negative_values}")
```

```python
# ==========================================================
# STEP 19 : Customers Dataset Cleaning
# Purpose:
# Verify missing values and duplicate customer IDs.
# ==========================================================

print("Missing Values")
display(customers.isnull().sum())

print("\nDuplicate Customer IDs")
print(customers["customer_id"].duplicated().sum())

print("\nDuplicate Customer Unique IDs")
print(customers["customer_unique_id"].duplicated().sum())
```

```python
# ==========================================================
# STEP 20 : Geolocation Dataset Cleaning
# Purpose:
# Remove duplicate latitude and longitude records.
# ==========================================================

print("Shape Before Removing Duplicates")
print(geolocation.shape)

geolocation = geolocation.drop_duplicates()

print("\nShape After Removing Duplicates")
print(geolocation.shape)
```

```python
# ==========================================================
# STEP 21 : Orders Dataset Cleaning
# Purpose:
# Check missing values in order timestamps.
# ==========================================================

print("Missing Values")
display(orders.isnull().sum())

print("\nDuplicate Order IDs")
print(orders["order_id"].duplicated().sum())
```

```python
# ==========================================================
# STEP 22 : Handle Missing Dates
# Purpose:
# Missing delivery dates usually indicate orders that
# were cancelled or not yet delivered.
# These values are retained instead of being removed.
# ==========================================================

date_columns = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

for column in date_columns:

    print(f"{column}")

    print(orders[column].isnull().sum())
```

```python
# ==========================================================
# STEP 23 : Order Items Dataset Cleaning
# Purpose:
# Verify missing values and duplicate order items.
# ==========================================================

print("Missing Values")
display(order_items.isnull().sum())

print("\nDuplicate Records")
print(order_items.duplicated().sum())
```

```python
# ==========================================================
# STEP 24 : Validate Price Columns
# Purpose:
# Check for zero or negative prices and freight values.
# ==========================================================

print("Zero Price Records")
print((order_items["price"] == 0).sum())

print("\nNegative Price Records")
print((order_items["price"] < 0).sum())

print("\nNegative Freight Charges")
print((order_items["freight_value"] < 0).sum())
```

```python
# ==========================================================
# STEP 265: Order Payments Dataset Cleaning
# Purpose:
# Check payment-related information.
# ==========================================================

print("Missing Values")
display(order_payments.isnull().sum())

print("\nNegative Payment Values")
print((order_payments["payment_value"] < 0).sum())

print("\nZero Payment Values")
print((order_payments["payment_value"] == 0).sum())
```

```python
# ==========================================================
# STEP 26 : Order Reviews Dataset Cleaning
# Purpose:
# Verify review scores and missing comments.
# ==========================================================

print("Missing Values")
display(order_reviews.isnull().sum())

print("\nReview Score Distribution")
display(order_reviews["review_score"].value_counts().sort_index())
```

```python
# ==========================================================
# STEP 27 : Products Dataset Cleaning
# Purpose:
# Identify missing product information.
# ==========================================================

print("Missing Values")
display(products.isnull().sum())

print("\nDuplicate Product IDs")
print(products["product_id"].duplicated().sum())
```

```python
# ==========================================================
# STEP 28 : Fill Missing Product Category
# Purpose:
# Replace missing categories with 'Unknown'.
# ==========================================================

products["product_category_name"] = products[
    "product_category_name"
].fillna("Unknown")

print("Missing Categories After Filling")
print(products["product_category_name"].isnull().sum())
```
